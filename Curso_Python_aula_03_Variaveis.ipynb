## 1. O que é uma variável?
Imagine o computador como um armário cheio de **caixas de papelão**. Cada caixa tem:

| Caixa | Python |
|---|---|
| Etiqueta | **nome** da variável |
| Conteúdo | **valor** guardado |
| Colocar algo na caixa | **atribuição** com o sinal `=` |

> ⚠️ Em Python, `=` **não** significa "igual a" como na matemática. Significa *"guarde o valor da direita na variável da esquerda"*. Para comparar valores usamos `==` (veremos nas próximas aulas).

```python
nome_do_cliente = "Andreza Souza Costa " # Cria uma váriável "usuário" e guarda o texto nome "Andreza"
print(nome_do_cliente)      # Mostra o valor guardado na variável
```

Não precisamos digitar "Andreza" de novo: basta usar o nome `usuario`. **Guardar uma informação para usar depois** é a grande vantagem das variáveis.

## 2. Os 4 tipos básicos de dados
O Python descobre o tipo **sozinho**, pelo valor que você atribui. Não é preciso declarar o tipo antes.

| Tipo | Nome | Guarda | Exemplo |
|---|---|---|---|
| `str` | String | texto (sempre entre aspas) | `"Programador Iniciante"` |
| `int` | Inteiro | números sem casas decimais | `42` |
| `float` | Ponto flutuante | números com casas decimais (use **ponto**, não vírgula) | `3.12` |
| `bool` | Booleano | verdadeiro ou falso | `True` / `False` |

```python
# TIPOS BÁSICOS DE VÁRIÁVEIS

# TIPO TEXTO (str/string)
nome_do_canal = 'Programador Iniciante'
print(nome_do_canal)
print(50 * "-")

# TIPO INTEIRO (int/integer)
videos_postados = 42
print(videos_postados)
print(50 * "-")

# TIPO DECIMAL (Float/ ponto Flutuante/ Numero decimal)
versão_python = 3.16
print(versão_python)
print(50 * "-")

# TIPO BOLEANO ( verdadeiro ? Falso /Boll)
curso_concluido = False
print(curso_concluido )
print(50 * "-")
```

### 🔍 Descobrindo o tipo com `type()`
A função `type()` mostra de que tipo é uma variável.

```python
print(nome_do_canal)
print(type(nome_do_canal))
print(50 * "-")
print(videos_postados)
print(type(videos_postados))
print(50 * "-")
print(versão_python)
print(type(versão_python))
print(50 * "-")
print(curso_concluido)
print(type(curso_concluido))
print(50 * "-")
```

> 🧠 **Dica:** `"42"` (com aspas) é **texto**; `42` (sem aspas) é **número**. Parecem iguais, mas o Python os trata de forma diferente!

```python
print(type('42'))
print(type(42))
```

## 3. Usando várias variáveis no `print()`
O `print()` aceita vários itens separados por **vírgula** e coloca um espaço entre eles. Assim misturamos texto fixo com os valores das variáveis.

```python
print("O meu Canal ",nome_do_canal, "tem", videos_postados, "videos postados.")
print("Estou utilizando a versao",versão_python, "do Python.")
print("Meu curso esta como concluido ?",curso_concluido )
```

## 4. Alterando o valor de uma variável
Variáveis podem **variar**: basta atribuir um novo valor. A partir daí, o Python usa o valor novo.

```python
videos_postados = 42
print("Antes", videos_postados)

videos_postados = 50
print("Depois", videos_postados)

print("Valor atual", videos_postados)
```

## 5. Regras para nomear variáveis

| Regra | ✅ Válido | ❌ Inválido |
|---|---|---|
| Use letras, números e `_` | `nome_cliente` | `nome-cliente` |
| Não comece com número | `cliente1` | `1cliente` |
| Sem espaços | `preco_produto` | `preco produto` |
| Maiúsculas ≠ minúsculas | `nome` e `Nome` são variáveis diferentes | |
| Não use palavras reservadas | `quantidade` | `if`, `for`, `class` |

**Boa prática:** escolha nomes que expliquem o conteúdo. Prefira `preco_produto` a `x`. Em Python o costume é usar letras minúsculas com `_` (*snake_case*).

## 6. Onde usamos variáveis na vida real?

- 🏠 **Automação:** temperatura, estado de sensores

Exemplo de loja virtual:

```python
# EXEMPLO : LOJA VIRTUAL

produto = "Camiseta"
preço = 59.90
estoque = 120
disponível = True

print("Produto", produto)
print("Preço", preço)
print("Estoque", estoque)
print("Disponivel", disponível)


```
