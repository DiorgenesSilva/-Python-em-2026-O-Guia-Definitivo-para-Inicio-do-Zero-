## 🧠 A História por Trás: George Boole, o Pai da Lógica Moderna

Antes de aprendermos os operadores na prática, vale parar 2 minutinhos para conhecer **de onde** tudo isso veio — porque essa é uma das histórias mais bonitas da computação.

Tudo começa com **George Boole** (1815–1864), um matemático, filósofo e lógico britânico. Em meados do século XIX (principalmente em sua obra de 1854, *"An Investigation of the Laws of Thought"*), Boole teve uma ideia revolucionária: e se a **lógica** — que até então era um assunto puramente filosófico, discutido desde Aristóteles — pudesse ser tratada como **matemática**?

![George Boole](https://commons.wikimedia.org/wiki/Special:FilePath/George%20Boole.jpg?width=220)

*George Boole (1815–1864), matemático e lógico britânico*

Foi exatamente isso que ele fez. Boole criou a **Álgebra Booleana**, reduzindo qualquer valor lógico a apenas **dois estados possíveis**:

> ✅ **Verdadeiro** ou ❌ **Falso**

Soa familiar? Deveria! É **exatamente** o `True` e `False` que vimos na Aula 04 e que estamos usando o tempo todo nesta aula. Na computação atual, esses dois estados também aparecem representados como os bits **1** e **0**.

### 🔗 A ligação direta com o que estamos estudando hoje

Tudo neste notebook é, no fundo, uma tradução direta do trabalho de Boole:

- **Operadores Lógicos (`and`, `or`, `not`)** são a tradução direta das operações que ele descreveu: `AND` (E) só é verdadeiro se tudo for verdadeiro; `OR` (OU) é verdadeiro se ao menos uma coisa for verdadeira; `NOT` (NÃO) inverte o valor.
- **Operadores de Comparação (`>`, `<`, `==`, `!=`)** — os símbolos de maior/menor já existiam antes dele (vieram de outros matemáticos, como Thomas Harriot) — mas é o trabalho de Boole que explica **por que** o resultado de `10 > 5` não é um texto, e sim um valor booleano: `True`.

## 🟦 PASSO 01: Operadores de Comparação

Operadores de comparação **comparam dois valores** e o resultado é **sempre** um booleano: `True` ou `False`.

| Operador | Significado            | Exemplo     | Resultado |
|----------|-------------------------|-------------|-----------|
| `==`     | Igual a                | `5 == 5`    | `True`    |
| `!=`     | Diferente de            | `5 != 3`    | `True`    |
| `>`      | Maior que               | `7 > 10`    | `False`   |
| `<`      | Menor que               | `7 < 10`    | `True`    |
| `>=`     | Maior ou igual a        | `5 >= 5`    | `True`    |
| `<=`     | Menor ou igual a        | `4 <= 3`    | `False`   |

⚠️ **Atenção ao clássico erro de principiante:**
- `=` → serve para **atribuir** um valor a uma variável (`idade = 18`)
- `==` → serve para **comparar** se dois valores são iguais (`idade == 18`)

Confundir esses dois é, de longe, um dos erros mais comuns de quem está começando!


````python
# Comparando números
print(10 == 10)   # True  -> são iguais
print(10 != 5)    # True  -> são diferentes
print(10 > 5)     # True  -> 10 é maior
print(10 < 5)     # False -> 10 não é menor
print(10 >= 10)   # True  -> são iguais (vale o "igual" também)
print(10 <= 9)    # False
````

````python
# Comparação não serve só para número! Também funciona com texto (string)
nome_digitado = "python"
senha_correta = "Python"

print(nome_digitado == senha_correta)        # False -> Python tem letra maiúscula diferente
print(nome_digitado == senha_correta.lower())# True  -> agora os dois estão minúsculos
````

💡 **Dica de Programador Experiente:** strings em Python são *case sensitive* (sensíveis a maiúsculas e minúsculas). `"Python"` e `"python"` são **diferentes** para o Python. É por isso que sites sempre avisam "senha incorreta" mesmo quando você digitou certo, só errando o Caps Lock!

## 🟦 PASSO 02: O Booleano por trás de tudo

Toda vez que você usa um operador de comparação, o Python devolve um valor do tipo `bool`. Podemos inclusive **guardar esse resultado em uma variável**, como fazemos com qualquer outro tipo:


````python
idade = 20
maior_de_idade = idade >= 18

print(maior_de_idade)        # True
print(type(maior_de_idade))  # <class 'bool'>
````

Isso é poderoso! Significa que podemos calcular uma condição **antes** e usar o resultado (a variável booleana) **depois**, deixando o código muito mais legível. Esse é um hábito de quem já tem um pouco mais de prática em programação.

## 🟦 PASSO 03: Operadores Lógicos (`and`, `or`, `not`)

E se precisarmos verificar **mais de uma condição ao mesmo tempo**? É aí que entram os operadores lógicos:

| Operador | O que faz                                              |
|----------|----------------------------------------------------------|
| `and`    | **TODAS** as condições precisam ser verdadeiras           |
| `or`     | **PELO MENOS UMA** condição precisa ser verdadeira         |
| `not`    | **INVERTE** o valor (True vira False, e vice-versa)        |


````python
idade = 25
tem_carteira_de_motorista = True

# AND -> as DUAS coisas precisam ser verdadeiras
pode_dirigir = idade >= 18 and tem_carteira_de_motorista
print("Pode dirigir?", pode_dirigir)  # True

# OR -> basta UMA ser verdadeira
fim_de_semana = False
feriado = True
pode_descansar = fim_de_semana or feriado
print("Pode descansar?", pode_descansar)  # True

# NOT -> inverte o resultado
esta_chovendo = False
print("Está sem chuva?", not esta_chovendo)  # True
````

## 🟦 PASSO 04: Tabela-Verdade — entendendo `and`, `or` e `not` por dentro

Para realmente **dominar** os operadores lógicos (e não só usar de forma intuitiva), é importante conhecer a tabela-verdade — ela mostra **todas** as combinações possíveis:

**`and` (E)** — só é `True` quando os dois lados são `True`

| A     | B     | A and B |
|-------|-------|---------|
| True  | True  | **True**|
| True  | False | False   |
| False | True  | False   |
| False | False | False   |

**`or` (OU)** — só é `False` quando os dois lados são `False`

| A     | B     | A or B  |
|-------|-------|---------|
| True  | True  | True    |
| True  | False | True    |
| False | True  | True    |
| False | False | **False**|

**`not` (NÃO)** — simplesmente inverte

| A     | not A |
|-------|-------|
| True  | False |
| False | True  |


````python
# Vamos provar a tabela-verdade na prática, testando as 4 combinações possíveis
print("Tabela do AND")
print(f"True  and True   ->  {True and True}")
print(f"True  and False  ->  {True and False}")
print(f"False and True   ->  {False and True}")
print(f"False and False  ->  {False and False}")

print("\nTabela do OR")
print(f"True  or True   ->  {True or True}")
print(f"True  or False  ->  {True or False}")
print(f"False or True   ->  {False or True}")
print(f"False or False  ->  {False or False}")
````

💡 **Dica de Programador Experiente:** memorizar a tabela-verdade de cabeça não é o objetivo. O importante é internalizar a ideia central:
- `and` é "exigente" (todo mundo tem que concordar)
- `or` é "tolerante" (basta um concordar)

Com essa intuição, você já resolve 90% dos casos do dia a dia sem nem precisar pensar na tabela.

## 🟦 PASSO 05: Combinando Comparação + Lógicos (o pulo do gato)

Na prática, quase nunca usamos um operador lógico isolado. O mais comum é **combinar comparações** usando `and`, `or` e `not` para criar regras mais elaboradas — exatamente como acontece em sistemas reais (login, formulários, regras de negócio, jogos).


````python
# Exemplo: sistema de inscrição em uma maratona
idade = 22
peso_kg = 68
exame_medico_aprovado = True

pode_correr_maratona = (idade >= 18 and idade <= 60) and peso_kg >= 50 and exame_medico_aprovado

print("Pode correr a maratona?", pode_correr_maratona)
````

````python
# Exemplo: liberação de desconto em uma loja
e_cliente_vip = False
valor_da_compra = 350
cupom_valido = True

tem_desconto = e_cliente_vip or (valor_da_compra >= 300 and cupom_valido)

print("O cliente ganhou desconto?", tem_desconto)
````

⚠️ **Atenção à ordem de prioridade (precedência):**

Assim como na matemática o `*` é resolvido antes do `+`, na lógica o `and` é resolvido **antes** do `or`. Isso pode gerar resultados inesperados se você não usar parênteses!


````python
# Sem parênteses -> Python decide a ordem por você (and primeiro)
resultado_1 = True or False and False
print(resultado_1)  # True -> primeiro calcula (False and False) = False, depois (True or False) = True

# Com parênteses -> você decide a ordem (e o código fica mais claro de ler)
resultado_2 = (True or False) and False
print(resultado_2)  # False -> primeiro calcula (True or False) = True, depois (True and False) = False
````

💡 **Dica de Programador Experiente:** mesmo quando você sabe a ordem de prioridade, **use parênteses** em expressões com `and` e `or` misturados. Não é "feio" nem redundante — é o tipo de cuidado que separa um código amador de um código profissional, porque deixa a intenção clara para qualquer pessoa (inclusive você mesmo, 6 meses depois) que for ler.

## 🟦 PASSO 06: Curto-circuito (*short-circuit*) — Python é "vagabundo" e isso é bom

Esse é um conceito mais avançado, mas que todo programador deveria conhecer desde já: o Python é **inteligente (e econômico)** ao avaliar `and` e `or`. Ele **para de calcular assim que já sabe a resposta final** — isso se chama **curto-circuito**.

- Em um `and`, se o **primeiro** valor já for `False`, o Python nem perde tempo checando o segundo — o resultado já é `False` de qualquer jeito.
- Em um `or`, se o **primeiro** valor já for `True`, o Python nem checa o segundo — o resultado já é `True` de qualquer jeito.


````python
quantidade_em_estoque = 0

# Se quantidade_em_estoque for 0, o lado esquerdo do "and" já é False
# e o Python NEM CHEGA A CALCULAR o lado direito (1000 / quantidade_em_estoque)
# Se ele calculasse, daria erro de divisão por zero!
tem_estoque_suficiente = quantidade_em_estoque > 0 and (1000 / quantidade_em_estoque) > 5
print("Tem estoque suficiente?", tem_estoque_suficiente)
````

Isso não é só curiosidade — é **extremamente útil** na prática. Um caso clássico é evitar erro ao acessar algo que pode não existir:


````python
lista_de_compras = []

# Se a lista estiver vazia, a segunda condição NUNCA chega a ser avaliada
# (e por isso não corre o risco de dar erro tentando acessar uma posição que não existe)
if len(lista_de_compras) > 0 and lista_de_compras[0] == "arroz":
    print("O primeiro item é arroz")
else:
    print("Lista vazia ou o primeiro item não é arroz")
````

## 🟦 PASSO 07: Mini-Projeto — Validador de Acesso a um Sistema

Vamos juntar **tudo** que vimos até aqui (comparação + lógicos + boas práticas) em um exemplo mais robusto, parecido com o que você encontraria em um sistema de verdade.


````python
print("=== VALIDADOR DE ACESSO - ACADEMIA PYTHON FIT ===\n")

# Dados do "usuário" (valores fixos para praticarmos a lógica de validação)
idade_usuario = 17
mensalidade_em_dia = True
aceitou_termos_de_uso = True
numero_de_tentativas_login = 2

# Regra 1: é maior de idade?
e_maior_de_idade = idade_usuario >= 18

# Regra 2: está em dia com a academia?
esta_regular = mensalidade_em_dia and aceitou_termos_de_uso

# Regra 3: não excedeu o limite de tentativas de login (máximo 3)
nao_foi_bloqueado = numero_de_tentativas_login < 3

# Resultado final: só libera acesso se TODAS as regras forem atendidas
acesso_liberado = e_maior_de_idade and esta_regular and nao_foi_bloqueado

print(f"Maior de idade?              {e_maior_de_idade}")
print(f"Está regular com a academia?  {esta_regular}")
print(f"Não foi bloqueado?            {nao_foi_bloqueado}")
print("-" * 45)
print(f"ACESSO LIBERADO?               {acesso_liberado}")

if not e_maior_de_idade:
    print("\n⚠️ Motivo do problema: usuário é menor de idade.")
````

Repare que usamos `if not e_maior_de_idade:` no final — mesmo sem ainda termos dado a aula completa sobre `if` (isso vem na Aula 07!), você já consegue ler e entender essa linha **só por já saber operadores lógicos**.
