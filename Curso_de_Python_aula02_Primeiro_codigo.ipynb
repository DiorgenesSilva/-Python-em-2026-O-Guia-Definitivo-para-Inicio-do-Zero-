**Explique o Erro :**

O erro acontece porque você tentou exibir um texto (uma string) sem colocá-lo entre aspas. No Python, textos precisam estar envolvidos por aspas simples (') ou duplas ("). Sem elas, o interpretador do Python acha que você está tentando usar nomes de variáveis ou comandos e acaba não entendendo o espaço entre as palavras, gerando o erro de sintaxe (SyntaxError)
